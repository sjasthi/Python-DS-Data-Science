## Worked Example: the `Book` class
Read this example carefully and **run it**. Your class will follow exactly the same pattern.

```python
#@title Worked example: Book (title, author, pages)

class Book:
    """A Book has a title, an author and a number of pages."""

    # constructor: runs automatically when we create a new Book object
    def __init__(self, title, author, pages):
        self.title = title       # instance variable 1
        self.author = author     # instance variable 2
        self.pages = pages       # instance variable 3

    # __repr__: unambiguous, for programmers.
    # Best practice: it looks like the code that would re-create the object.
    def __repr__(self):
        return f"Book(title={self.title!r}, author={self.author!r}, pages={self.pages!r})"

    # __str__: friendly, for users. print() uses this one.
    def __str__(self):
        return f"'{self.title}' by {self.author} ({self.pages} pages)"


# create three objects
book_1 = Book("Python 101", "Learn and Help", 120)
book_2 = Book("Data Science Basics", "A. Kumar", 250)
book_3 = Book("Hello, Classes!", "B. Rao", 80)

print(book_1)          # calls __str__
print(repr(book_1))    # calls __repr__
print([book_1, book_2, book_3])   # a list shows each object with __repr__
```

This cell looks at the class you wrote and the objects you created, and tells you what still needs work.

A clean self-check does **not** guarantee full marks – your instructor also reads your code and output.

```python
#@title Self-check helper (run this cell once)
import inspect

EXPECTED_VARIABLES = {
    "Planet": ['name', 'diameter_km', 'num_moons'],
    "Recipe": ['title', 'cuisine', 'prep_minutes'],
    "VideoGame": ['title', 'genre', 'rating'],
    "Bicycle": ['brand', 'gear_count', 'color'],
    "Flight": ['flight_number', 'destination', 'departure_time'],
    "Plant": ['common_name', 'species', 'water_per_week'],
    "Sport": ['name', 'team_size', 'equipment'],
    "Instrument": ['name', 'family', 'price'],
    "Superhero": ['hero_name', 'power', 'city'],
    "Library": ['name', 'city', 'num_books'],
    "Painting": ['title', 'artist', 'year'],
    "Festival": ['name', 'month', 'region'],
    "Shoe": ['brand', 'size', 'price'],
    "Volcano": ['name', 'country', 'height_m'],
    "Pizza": ['size', 'crust', 'num_toppings'],
}

def check_my_class(cls):
    """Checks the lab requirements for the class you pass in."""
    passed = 0
    total = 7

    def report(ok, message):
        nonlocal passed
        passed += ok
        print(("✅ " if ok else "❌ ") + message)

    print(f"Checking class: {cls.__name__}\n" + "-" * 45)

    chosen = globals().get("selected_class", "?")
    report(cls.__name__ == chosen, f"Class name matches your selection ('{chosen}')")
    report(bool(cls.__doc__) and "TODO" not in cls.__doc__, "Class has a docstring")

    params = [p for p in inspect.signature(cls.__init__).parameters if p != "self"]
    report("__init__" in cls.__dict__ and len(params) >= 3,
           f"__init__ takes 3 parameters (found {len(params)}: {params})")

    objects = [v for v in globals().values() if isinstance(v, cls)]
    report(len(objects) >= 3, f"At least 3 objects created (found {len(objects)})")

    expected = EXPECTED_VARIABLES.get(cls.__name__, [])
    names_ok = bool(objects) and bool(expected) and all(set(expected) <= set(vars(o)) for o in objects)
    report(names_ok, f"Objects use the listed instance variables {expected}")

    repr_ok, str_ok = "__repr__" in cls.__dict__, "__str__" in cls.__dict__
    for obj in objects:
        n_vars = len(vars(obj))
        r, s = repr(obj) if repr_ok else None, str(obj) if str_ok else None
        if n_vars < 3:
            print(f"   ⚠️  an object has only {n_vars} instance variable(s)")
        if not isinstance(r, str) or not r.startswith(cls.__name__ + "("):
            repr_ok = False
        if not isinstance(s, str) or s == r:
            str_ok = False
    report(repr_ok and len(objects) > 0, "__repr__ returns a string like  ClassName(...)")
    report(str_ok and len(objects) > 0, "__str__ returns a friendly string (different from __repr__)")

    print("-" * 45 + f"\nSelf-check: {passed}/{total} checks passed")
```
